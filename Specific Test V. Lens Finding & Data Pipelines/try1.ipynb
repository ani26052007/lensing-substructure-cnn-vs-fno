```python
import numpy as np
import matplotlib.pyplot as plt

file_path = "/home/groot/AGV/ML4SCI/DeepLense/Specific Test V. Lens Finding & Data Pipelines/train_lenses/2.npy"
data = np.load(file_path)

# --- show channels ---
fig, axes = plt.subplots(1, 3, figsize=(12,4))
for i in range(3):
    axes[i].imshow(data[i], cmap='gray')   # 👈 cleaner indexing
    axes[i].set_title(f"Channel {i}")
    axes[i].axis('off')
plt.show()

# --- convert to RGB format ---
rgb = np.transpose(data, (1, 2, 0))

# normalize (VERY important)
rgb = (rgb - rgb.min()) / (rgb.max() - rgb.min())

# --- show RGB ---
plt.imshow(rgb)
plt.title("RGB Visualization")
plt.axis('off')
plt.show()
```
